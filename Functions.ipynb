```python
#Function that expects an unknown number of arguments
def show_args(*args):
    for i in args:
        print(i)
show_args(10,20,30)
```

```python
#Function that expects keyword arguments (kwargs)
def show_details(**kwargs):
    for key, value in kwargs.items():
        print(key,":",value)
show_details(name="Archish", age=20, course="B.Tech")
```

```python
#for prime number
def prime(num):
    if num <= 1:
        print("Not Prime")
    
    for i in range(2,num):
        if num%i==0:
            print("Not Prime")
    print("Prime") 

prime(13)

```

```python
# for prime number in a list

def prime(*numbers):
    for num in numbers:
        if num <= 1:
            print(num, "is Not Prime")
            continue

        for i in range(2, num):
            if num % i == 0:
                print(num, "is Not Prime")
                break
        else:
            print(num, "is Prime")


prime(1, 5, 8, 13)

```

```python
#Function to check whether a number is prime
def unnique(lst):
    result=[]
    for i in lst:
        if i not in result:
            result.append(i)
    return result
print(unnique([1,2,2,5,4]))
```

```python
#Palindrome
def is_palindrome(s):
    s = s.upper()
    if s == s[::-1]:

        print("Plaindrome")
    else:
        print("Not Palindrome")
is_palindrome("1551")
is_palindrome("Madam")
```

```python
#Count uppercase and lowercase letters in a string
def count_case(s):
    upper = 0
    lower = 0

    for i in s:
        if i.isupper():
            upper+=1
        elif i.islower():
            lower+=1

    print("Uppercase letters:", upper)
    print("Lowercase letters:", lower)

count_case("Hello World")
```

```python
# panmagram = contains all the 26 letters 
def is_pangram(s):
    s = s.lower()
    for ch in "qwertyuiopasdfghjklzxcvbnm":
        if ch not in s:
            return False
    return True

sentence = "The quick brown fox jumps over the lazy dog"
print(is_pangram(sentence))
```

```python
def is_anagram(s1, s2):
    if sorted(s1) == sorted(s2):
        return True
    else:
        return False 

print(is_anagram("listen", "silent"))
```
